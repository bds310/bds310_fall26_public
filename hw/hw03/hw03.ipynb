## Question 1: For Loops (14 pts)

For loops allow you to execute an action repeatedly by iterating through a sequence -- which can be a string, list, array, or other data type.

The general structure of a for loop is:

```
for temporary variable in iterable object:
       action
```
\
`temporary variable`:
- One item within the iterable object, updated each loop iteration.

`iterable object`:
- The sequence traversed during the loop.

`action`:
- The code that is executed each loop iteration.

### Relative and Absolute Paths

**Absolute Paths** are paths specified from the root of the filesystem
- ex. `/Users/billy/myproject/data/mydata.txt` or `~/myproject/data/mydata.txt`

Absolute paths are useful because they will be valid no matter where you are on your filesystem. But when explicitly typed out, absolute paths will fail on other filesystems, simply because other filesystems won't have the same directory structure.
For example, when Billy shares `myproject/` with his friend Tommy, Tommy will encounter an error telling him that `/Users/billy/` is not a valid directory.

**Relative Paths** are paths specified from your current directory
- ex. `data/mydata.txt` *assuming you are in* `/Users/billy/myproject/`

Relative paths avoid this issue (assuming working directories are standardized across filesystems). For example, even if Tommy puts `myproject/` in `/Users/tommy/Downloads/`, the path above will still be valid if executed from `myproject/`. However, relative paths fail if working directory structure is not standardized across filesystems or commands are executed from outside of the working directory.

Whether you should be using absolute or relative paths is context-dependent. **<span style="color:red">For this notebook, use either relative paths, or dynamic absolute paths (see below).</span>** Explicitly typed out absolute paths will fail when we attempt to run your submitted notebooks for grading.



````python
%%bash
# this cell sets up a dynamic symlink
# that will enable you to use absolute paths
# without causing issues on other filesystems
# don't alter this cell please

# remove old symlink (if one exists)
if [ -L ~/symhw01 ]; then
    echo "removing old symlink"
    rm ~/symhw01;
fi

# create symlink in the home directory called `symhw01` that
# points to the current directory (accessed with built in variable $PWD)
HW01_PATH=$PWD
ln -s $HW01_PATH ~/symhw01
````

````python
%%bash
# now we have flexibility when specifying paths
# don't alter this cell please

cat ~/symhw01/test.txt # we can use our handy symlink
echo ' (using symlink)'
cat test.txt # we can use relative paths
echo ' (using relative path)'
````
